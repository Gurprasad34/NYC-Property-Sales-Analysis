```python
import os
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv
```

```python
# Load database credentials
load_dotenv("../.env")

db_user = os.getenv("DB_USER")
db_password = os.getenv("DB_PASSWORD")
db_host = os.getenv("DB_HOST")
db_port = os.getenv("DB_PORT")
db_name = os.getenv("DB_NAME")

# Connect to MySQL
engine = create_engine(
    f"mysql+pymysql://{db_user}:{db_password}@{db_host}:{db_port}/{db_name}"
)
```

```python
# Load property sales data from MySQL
query = """
SELECT *
FROM property_sales;
"""

df = pd.read_sql(query, engine)

df.head()
```

```python
# Check duplicate rows
df.duplicated().sum()
```

```python
# Check empty strings
(df == "").sum()
```

```python
# Check important numeric columns
numeric_columns = [
    "sale_price",
    "residential_units",
    "commercial_units",
    "total_units",
    "land_square_feet",
    "gross_square_feet",
    "year_built"
]

df[numeric_columns].head(10)
```

```python
# Convert numeric columns
numeric_columns = [
    "borough",
    "block",
    "lot",
    "zip_code",
    "residential_units",
    "commercial_units",
    "total_units",
    "land_square_feet",
    "gross_square_feet",
    "year_built",
    "sale_price"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")
```

```python
# Convert sale date to datetime
df["sale_date"] = pd.to_datetime(
    df["sale_date"],
    errors="coerce"
)
```

```python
# Check our updated data types
df.dtypes
```

```python
# Check sale price distribution
df["sale_price"].describe()
```

```python
# Check zero-dollar sales
(df["sale_price"] == 0).sum()
```

```python
# Check low sale prices
df[df["sale_price"] < 1000]["sale_price"].value_counts().sort_index()
```

```python
# Look at low-price transactions
df.loc[
    df["sale_price"] < 1000,
    [
        "borough",
        "neighborhood",
        "building_class_category",
        "address",
        "sale_price",
        "sale_date"
    ]
].head(20)
```

```python
# Keep transactions with meaningful sale prices
df_clean = df[df["sale_price"] >= 1000].copy()
```

```python
# Check how many rows remain
print("Original rows:", len(df))
print("Clean rows:", len(df_clean))
print("Rows removed:", len(df) - len(df_clean))
```

```python
# Replace borough codes with borough names
borough_map = {
    1: "Manhattan",
    2: "Bronx",
    3: "Brooklyn",
    4: "Queens",
    5: "Staten Island"
}

df_clean["borough"] = df_clean["borough"].map(borough_map)
```

```python
# Check borough names
df_clean["borough"].value_counts()
```

```python
# Check the oldest building years
df_clean["year_built"].value_counts().sort_index().head(20)
```

```python
# Replace invalid year built values with missing values
df_clean.loc[df_clean["year_built"] < 1800, "year_built"] = pd.NA

# Check the minimum year built
df_clean["year_built"].min()
```

```python
# Check zero square footage
(df_clean["gross_square_feet"] == 0).sum()
```

```python
# Check smallest square footage values
df_clean["gross_square_feet"].value_counts().sort_index().head(20)
```

```python
# Replace zero square footage with missing values
df_clean.loc[df_clean["gross_square_feet"] == 0, "gross_square_feet"] = pd.NA

# Check that zero square footage values are gone
(df_clean["gross_square_feet"] == 0).sum()
```

```python
# Check for negative unit values
print("Residential:", (df_clean["residential_units"] < 0).sum())
print("Commercial:", (df_clean["commercial_units"] < 0).sum())
print("Total:", (df_clean["total_units"] < 0).sum())
```

```python
# Check zero land square footage
(df_clean["land_square_feet"] == 0).sum()
```

```python
# Replace zero land square footage with missing values
df_clean.loc[df_clean["land_square_feet"] == 0, "land_square_feet"] = pd.NA

# Check that zero values are gone
(df_clean["land_square_feet"] == 0).sum()
```

```python
# Create price per square foot
df_clean["price_per_sqft"] = (
    df_clean["sale_price"] / df_clean["gross_square_feet"]
)
```

```python
df_clean["price_per_sqft"].head(10)
```

```python
# Create date features
df_clean["sale_year"] = df_clean["sale_date"].dt.year
df_clean["sale_month"] = df_clean["sale_date"].dt.month
df_clean["sale_quarter"] = df_clean["sale_date"].dt.quarter
```

```python
# Check date features
df_clean[
    ["sale_date", "sale_year", "sale_month", "sale_quarter"]
].head()
```

```python
# Create property age
df_clean["property_age"] = (
    df_clean["sale_year"] - df_clean["year_built"]
)
```

```python
# Check property age
df_clean[
    ["year_built", "sale_year", "property_age"]
].head(10)
```

```python
# Check final dataset shape
print("Original shape:", df.shape)
print("Clean shape:", df_clean.shape)
```

```python
# Check final missing values
# Some nulls are expected because certain fields do not apply to every property.
# Square footage and unit information are not reported for many transactions.
# These rows are kept because they can still be used for sale price, location, and other analyses.
df_clean.isna().sum()
```

```python
# Check final data types
df_clean.dtypes
```

```python
# Save cleaned dataset
df_clean.to_csv(
    "../data/processed/property_sales_clean.csv",
    index=False
)
```
